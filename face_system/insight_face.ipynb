```python
'''def extract_embeddings(image_path):

    img = cv2.imread(image_path)

    faces = app.get(img)

    embeddings = []

    for face in faces:
        embeddings.append(face.embedding)

    return embeddings'''
def extract_embeddings(image_path):

    img = cv2.imread(image_path)

    faces = app.get(img)

    embeddings = []

    for face in faces:

        embedding = face.embedding

        # Normalize embedding
        embedding = embedding / np.linalg.norm(embedding)

        embeddings.append(embedding)

    return embeddings
```

```python
def store_face(image_path, latitude, longitude):

    embeddings = extract_embeddings(image_path)

    for emb in embeddings:

        face_data = {
            "id": str(uuid.uuid4()),
            "embedding": emb,
            "latitude": latitude,
            "longitude": longitude,
            "image_path": image_path
        }

        face_database.append(face_data)

    print("Faces stored:", len(embeddings))
```

```python
def cosine_similarity(a, b):

    return np.dot(a,b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
def search_face(image_path, threshold=0.5):

    embeddings = extract_embeddings(image_path)

    if len(embeddings) == 0:
        print("No face detected")
        return []

    matches = []

    # Loop through all detected faces in query image
    for query_embedding in embeddings:

        for face in face_database:

            sim = cosine_similarity(query_embedding, face["embedding"])

            if sim > threshold:

                matches.append({
                    "similarity": float(sim),
                    "latitude": face["latitude"],
                    "longitude": face["longitude"],
                    "image_path": face["image_path"]
                })
                break
    return matches
```

```python
def extract_and_crop_faces(image_path):

    img = cv2.imread(image_path)

    faces = app.get(img)

    cropped_faces = []

    for i, face in enumerate(faces):

        bbox = face.bbox.astype(int)

        x1, y1, x2, y2 = bbox

        cropped = img[y1:y2, x1:x2]

        cropped_faces.append(cropped)

        cv2.imwrite(f"face_{i}.jpg", cropped)

    return cropped_faces
```

```python
# Store face with embeddings to Supabase (image + metadata)
def store_face_to_supabase(image_path, latitude, longitude):
    """Extract face embeddings and store both image and embeddings in Supabase"""
    
    # Read image
    with open(image_path, 'rb') as f:
        image_data = f.read()
    
    # Extract embeddings
    embeddings = extract_embeddings(image_path)
    
    if len(embeddings) == 0:
        print("No faces detected in image")
        return []
    
    stored_faces = []
    
    for idx, emb in enumerate(embeddings):
        # Generate unique ID for this face
        face_id = str(uuid.uuid4())
        
        # Upload image to Supabase storage
        image_name = f"{face_id}_{idx}.jpg"
        supabase.storage.from_("faces").upload(image_name, image_data)
        image_url = supabase.storage.from_("faces").get_public_url(image_name)
        
        # Store embedding metadata in database
        face_record = {
            "id": face_id,
            "embedding": emb.tolist(),
            "latitude": latitude,
            "longitude": longitude,
            "image_url": image_url,
            #"image_name": image_name
        }
        
        supabase.table("faces").insert(face_record).execute()
        stored_faces.append(face_record)
        print(f"Stored face {idx + 1} with ID: {face_id}")
    
    return stored_faces
```

```python
# Load all faces from Supabase into memory for search
def load_faces_from_supabase():
    """Fetch all stored face embeddings from Supabase database"""
    
    try:
        response = supabase.table("faces").select("*").execute()
        faces = response.data
        print(f"Loaded {len(faces)} faces from Supabase")
        return faces
    except Exception as e:
        print(f"Error loading faces: {e}")
        return []

# Load faces when notebook starts
db_faces = load_faces_from_supabase()
print(f"Database ready with {len(db_faces)} faces")
```

```python
# Search for similar faces in Supabase database
import ast


def search_face_in_supabase(image_path, threshold=0.45):
    """Search for matching faces in Supabase database"""
    
    embeddings = extract_embeddings(image_path)
    
    if len(embeddings) == 0:
        print("No face detected in query image")
        return []
    
    matches = []
    
    # Reload fresh data for accurate search
    db_faces = load_faces_from_supabase()
    
    for query_idx, query_emb in enumerate(embeddings):
        print(f"\nSearching for face {query_idx + 1}...")
        
        for stored_face in db_faces:
            
            

            stored_emb = np.array(ast.literal_eval(stored_face["embedding"]), dtype=np.float32)
            sim = cosine_similarity(query_emb, stored_emb)
            if sim > threshold:
                match = {
                    "query_face_id": query_idx,
                    "stored_face_id": stored_face["id"],
                    "similarity": float(sim),
                    "latitude": stored_face["latitude"],
                    "longitude": stored_face["longitude"],
                    "image_url": stored_face["image_url"]
                }
                matches.append(match)
                print(f"  ✓ Match found! Similarity: {sim:.4f}")
    
    print(f"\nTotal matches found: {len(matches)}")
    return matches

# Test search
test_results = search_face_in_supabase("photo_bts_face_recog.jpg", threshold=0.45)
print(test_results)
```
